```python
def get_races_for_year(year):
    races = []

    for rnd in range(1, 30):  
        try:
            session = fastf1.get_session(year, rnd, "R")  
            session.load()
            races.append({
                "Year": year,
                "Round": rnd,
                "RaceName": session.event.EventName
            })
        except:
            pass

    return races

races = get_races_for_year(2022)

print(races)
```

```python
sessions_df = pd.read_csv("race_f1_sessions25.csv")
sessions_df["Round"].max()
```

```python
for round, compound_dict in compound_map_2025.items():
    sessions_df.drop(
        sessions_df[(sessions_df['Round'] == round) & (sessions_df['TrackStatus'].isin([4,5,6,7]))].index,
        inplace=True
    )
    mask = sessions_df['Round'] == round


    for old_c, new_c in compound_dict.items():
        mask = (sessions_df["Round"] == round) & (sessions_df['Compound'] == old_c)
        sessions_df.loc[mask, 'Compound'] = new_c  
```

```python
mask = (sessions_df['Compound'] == "INTERMEDIATE")
sessions_df.loc[mask, 'Compound'] = 'I'  

mask = (sessions_df['Compound'] == "WET")
sessions_df.loc[mask, 'Compound'] = 'W'  
```
